# بررسی محیط پروژه تشخیص سبک موسیقی
این نوت‌بوک نسخه Python، مسیر مفسر و نسخهٔ کتابخانه‌های پروژه را در همان kernel ثبت می‌کند. همهٔ سلول‌ها را به‌ترتیب اجرا کنید و فایل را با خروجی‌های واقعی ذخیره کنید.

```python
import sys, platform, importlib.metadata as metadata
print("Python:", sys.version)
print("Executable:", sys.executable)
print("Platform:", platform.platform())
```

```python
packages = ["numpy", "matplotlib", "librosa", "scikit-learn", "jupyter"]
for package in packages:
    try:
        print(f"{package}: {metadata.version(package)}")
    except metadata.PackageNotFoundError:
        print(f"{package}: NOT INSTALLED")
```

```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
print("NumPy:", np.__version__)
print("Matplotlib:", matplotlib.__version__)
X = np.array([[0, 0], [1, 1], [4, 4], [5, 5]])
y = ["genre_a", "genre_a", "genre_b", "genre_b"]
model = make_pipeline(StandardScaler(), LogisticRegression()).fit(X, y)
print("Smoke-test prediction:", model.predict([[0.2, 0.1]])[0])
plt.bar(["genre_a", "genre_b"], [2, 2])
plt.title("Environment smoke test")
plt.ylabel("toy sample count")
plt.show()
```
